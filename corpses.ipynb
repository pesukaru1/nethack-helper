```python
import os
from bs4 import BeautifulSoup
import re
import pandas as pd
from io import StringIO

HTMLFILE = 'corpses.html'
RESTYPE = 'fire'
resist_pattern = rf'\b{RESTYPE.title()}\b:\s\d*%'

if not os.path.exists(HTMLFILE):
    print('File does not exist')

with open(HTMLFILE, 'r', encoding='utf-8') as file:
    html = file.read()  # читает весь файл как строку
    soup = BeautifulSoup(html, "lxml")

corpses_table = soup.find('table')
pd_table = pd.read_html(StringIO(str(corpses_table)))[0]
pd_table.columns

mask = pd_table['Intrinsics'].str.contains(resist_pattern, regex=True, na=False)
result = pd_table[mask]
result['Name']
```
